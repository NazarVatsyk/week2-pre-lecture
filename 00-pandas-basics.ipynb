## 1. How a notebook works

A notebook is a list of cells. A cell holds either text, like this one, or code, like
the next one. You run a code cell with **Shift+Enter**: the code runs, its result
appears under the cell, and the cursor moves to the next cell.

```python
1 + 1
```

A code cell shows something under it when it runs. Which values?

```python
# Predict: what appears under this cell — 10, 15, or both?
x = 5
x * 2
x * 3
#15
```

<details>
<summary>Answer</summary>

A code cell shows the value of its **last line** and nothing else. Only 15. `x = 5` is an assignment and has no value to show, `x * 2` is not the last
line, and `x * 3` is.

</details>

To show more than one thing, use `print`:

```python
print(x * 2)
print(x * 3)
```

### Cells share one memory

Every cell runs in the same Python session, called the **kernel**. A variable made in
one cell exists for every cell you run afterwards, wherever that cell sits on the page.
The number in brackets on the left of a code cell says when it ran.

```python
y = x + 1
y
```

page.

## 2. A Series is a list of values with a label on each one

pandas is imported under the short name `pd`.

```python
import pandas as pd

print(pd.__version__)
```

```python
a = pd.Series([1, 2, 3, 10, 3, 7])
a
```

The left column is the **index**: one label per value. When you do not give labels,
pandas numbers the values from 0. The last line names the type of the values.

You can give your own labels:

```python
a = pd.Series([1, 2, 3, 10, 3, 7], index=["a", "b", "c", "d", "e", "f"])
a
```

Now a Series works like a dictionary: a label finds a value.

```python
a["d"]
```

### pandas and NumPy

The value came back as `np.int64(10)`, not as `10`. `np` is NumPy, a library for arrays
of numbers. pandas does not store numbers itself: it reuses NumPy's objects and types, the
way one library can build on another's work. `int64` is NumPy's type for a whole number
kept in a 64-bit slot.

Python's own `1` and NumPy's `np.int64(1)` are both the number one, and they compare
equal. A Python `int` is a separate object that can grow as large as memory allows. A
NumPy `int64` is one fixed-size slot in a block where every value has the same type,
which is what lets a whole column be processed in one call. In your code the two mix
freely, and `print` shows the bare number for either.

```python
value = a["d"]
print(value == 10)
print(type(value))
print(type(10))
```

```python
# Predict: what happens when there are three values and two labels?
pd.Series([1, 2, 3], index=["a", "b"])
```

<details>
<summary>Answer</summary>

A `ValueError`. The index must have exactly one label per value.

</details>

## 3. Arithmetic works label by label

Make a second Series with some labels in common with `a` and some new ones:

```python
b = pd.Series([1, 5, 2, 4], index=["a", "c", "g", "x"])
b
```

```python
# Predict: how many values will the result have, and which ones will be numbers?
a + b
```

<details>
<summary>Answer</summary>

Eight values, and only two of them are numbers. pandas adds label by label: `a` at label
`a` plus `b` at label `a` gives 2, and the same for `c`. Where a label exists on one side
only, there is nothing to add, so the result is `NaN`.

</details>

`NaN` stands for "not a number" and is how pandas marks a missing value.

This matching by label is called **alignment**. It happens in every operation between
two Series. Arithmetic with a single number needs no matching:

```python
a + 7
```

```python
a * 2
```

### pandas against plain Python

The same `+` means something else on a Python list. Two lists are joined end to end:

```python
[1, 2, 3] + [10, 20]
```

A list plus a number is an error, because Python does not know whether you mean every
element. pandas defines both: `+` between two Series adds label by label, and `+` with a
number adds it to every value.

**Try it.** In the cell below, change one label in `b2` to `"d"` and run it. How many
numbers now?

```python
# Try it
b2 = pd.Series([1, 5, 2, 4], index=["a", "c", "d", "x"])
a + b2
```

## 5. A table from a file

The table is one month of sales from an online shop, 78 015 rows. One row is one line on one receipt: one product,

The table is read straight from the web.

```python
url = "https://raw.githubusercontent.com/stat163-2026t1/week2-pre-lecture/main/data/online_retail_2010_11.csv"
df = pd.read_csv(url)
df.head()
```

/`head()` shows the first five rows. `shape` gives (rows, columns), and `columns` lists

There are two ways to reach a column: `df["Quantity"]` and `df.Quantity`. Only the first
works for every name. A name with a space or a dash, such as `Customer ID`, or a name
pandas already uses for a method, such as `count`, cannot follow the dot. This course
uses brackets.

```python
df.shape
```

```python
df.columns
```

```python
df["Quantity"]
```

A column is a Series, so everything from parts 2 and 3 applies to it. Two questions a
column can answer:

```python
print(df["Quantity"].max())
print(df["Quantity"].mean())
```

### Why pandas, and not a loop

You could compute the same things with a Python loop. Here is quantity times price for
each line, both ways. `%%time` at the top of a cell reports how long the cell
took.

```python
%%time
revenue = []
for i in range(len(df)):
    revenue.append(df["Quantity"].iloc[i] * df["Price"].iloc[i])
```

```python
%%time
revenue = df["Quantity"] * df["Price"]
```

The gap grows with the table. The reason is how the values are stored. In plain Python every number is a separate object with its own
type attached, and a loop handles them one at a time: read the type, do the arithmetic,
make a new object, and repeat 78 015 times. A Series keeps its values in one block of
memory with one type for the whole column, and an operation on it runs once, in compiled
code written in C, over the whole block. That is also why a column has exactly one
storage type: one type per block.

## 6. A condition gives a boolean mask

Compare a column with a number and you get a new Series, one `True` or `False` per row,
with the same labels. The comparison you write is the **condition**; the Series it
produces is the **boolean mask**, or mask for short. The picture: a mask lies over the
table and lets a row through where it is `True`.

```python
df["Quantity"] > 20
```

In plain Python, `5 > 20` gives one answer, and a list compared with a number is an
error. pandas defines the comparison per value, so one line asks the question of every
row at once.

```python
# Predict: what does this number count?
(df["Quantity"] > 20).sum()
```

<details>
<summary>Answer</summary>

`True` counts as 1 and `False` as 0, so the sum of a mask is the number of rows where
the condition holds. This is the usual way to count things in pandas.

</details>

Put the mask inside `df[...]` and you keep only the rows where it is `True`:

```python
df[df["Quantity"] > 20]
```

Look at the index on the left. The labels jump: 3, 7, 8, 11. Each row kept the label it
had in the full table. Filtering removes rows and changes nothing else.

So square brackets mean two things. A column **name** inside them gives a column. A
**mask** inside them gives rows.

### Combining conditions

In a loop you would write `and`, one row at a time:

```python
matches = 0
for q in df["Quantity"]:
    if q >= 19 and q <= 25:
        matches = matches + 1
matches
```

On whole columns `and` does not work. Python's `and` needs one `True` or `False` to
decide, and a mask holds 78 015 of them, one per row, so Python stops with an error
about the truth value of a Series being ambiguous. pandas gives three operators that
work value by value instead: `&` means and, `|` means or, `~` means not. Every condition
goes in its own parentheses, because Python does `&` before `>=`, the opposite of the
order it uses for `and`.

```python
between = df[(df["Quantity"] >= 19) & (df["Quantity"] <= 25)]
between.shape
```

```python
# Predict: what happens without the parentheses?
df[df["Quantity"] >= 19 & df["Quantity"] <= 25]
```

<details>
<summary>Answer</summary>

A `ValueError` with a long message about "the truth value of a Series", the same error
that `and` gives. Python reads `&` before `>=`, so it tries to combine the number 19 with
the column first, and the rest fails.

</details>

The same rows, written the other way round: keep everything that is **not** below 19
**or** above 25.

```python
outside = (df["Quantity"] < 19) | (df["Quantity"] > 25)
df[~outside].shape
```

The two shapes are equal, because the two masks pick the same rows.

You can also filter in two steps:

```python
step1 = df[df["Quantity"] >= 19]

# Predict: is the shape the same as `between`? Look closely at the second line.
step2 = step1[step1["Quantity"] <= 25]
step2.shape
```

<details>
<summary>Answer</summary>

The same count, and a warning. The second line filters `step1` with a mask computed
on `df`. The two have different rows, so pandas lined them up by label before filtering,
which is alignment again, and then warned that it had to. Here that gives the right
answer. With labels that do not match, it gives a wrong one without any warning you would
notice. Compute the mask on the table you filter: `step1[step1["Quantity"] <= 25]`.

</details>

**Try it.** Count the rows shipped to France with a quantity above 50. You need a
condition on `Country`, a condition on `Quantity`, `&` between them, and `.sum()` at
the end.

```python
import numpy as np
df.columns
df["Quantity"] = np.int64(df["Quantity"])
len(df[(df["Country"]=="France")&(df["Quantity"]>50)])# Try it
```

## 7. Picking by label and by position

Two ways to point at one row or one value:

- `.loc[label]` uses the **label** from the index.
- `.iloc[position]` uses the **position**, counting from 0, whatever the label says.

A Python list knows only positions, `items[0]`. A dictionary knows only keys, `d["a"]`.
A Series is both at once, so it needs two doors. On the small Series from part 2:

```python
a
```

```python
print(a.loc["e"])
print(a.iloc[3])
```

Remove one value and see which of the two changes:

```python
a2 = a.drop("d")
a2
```

`drop` gave back a new Series and left `a` unchanged. Almost every pandas method works
this way: it returns a new object and touches nothing. To keep a result, assign it to a
name.

```python
# Predict: which of the two answers changes, compared with the cell above?
print(a2.loc["e"])
print(a2.iloc[3])
```

<details>
<summary>Answer</summary>

`.loc["e"]` still gives 3: the label `e` still points at the same value. `.iloc[3]` now
gives 3 too, but for a different reason: the value 10 is gone, so the fourth position
holds what used to be the fifth. A label follows its value. A position counts places.

</details>

On the big table the labels are numbers, so the two look alike until rows are removed
or reordered:

```python
df.loc[102]
```

```python
df.iloc[1]
```

### Ranges

Both accept a range with a colon.

```python
# Predict: how many values does each line return?
print(len(a.loc["c":"f"]))
print(len(a.iloc[1:3]))
```

<details>
<summary>Answer</summary>

`.loc` **includes** the end label. `.iloc` **excludes** the end position, like a Python
list. `"c":"f"` takes `c`, `d`, `e`, `f`. `1:3` takes positions 1 and 2.

</details>

`.loc` with a range follows the order the values are in right now. Sort the Series and
the same range gives a different answer:

```python
b_sorted = b.sort_values()
b_sorted
```

```python
# Predict: how many values? Look at where "c" and "g" are now.
b_sorted.loc["c":"g"]
```

<details>
<summary>Answer</summary>

Nothing: an empty Series. After sorting, `g` comes before `c`, and a range does not run backwards. The same
line on the unsorted `b` gives two values, `c` and `g`. A label range depends on the current order
of the rows, so use it only when you know that order.

</details>

## Summary

- Cells share one memory. The order you ran them in is what counts. Restart and run

- A Series is values with labels. A DataFrame is Series side by side, sharing one index.
